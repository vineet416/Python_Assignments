```python
name = "Vineet Patel"
name[::-1]
```

# 2. Write a code to count the number of vowels in a string.

```python
name = "Vineet Patel"
vowels = ["a", "e", "i", "o", "u"]
vowels_count = 0

for char in name:
    if char.lower() in vowels:
        vowels_count += 1

vowels_count
```

# 3. Write a code to check if a given string is a palindrome or not

```python
word = input("Enter a word")

if word.lower() == word.lower()[::-1]:
    print("word is a Palindrome")
else:
    print("word is not Palindrome")
```

# 4. Write a code to check if two given string are anagrams of each other.

```python
string_1 = input("Enter first word").lower()
string_2 = input("Enter second word").lower()

if len(string_1) != len(string_2):
    print(f"{string_1} and {string_2} are not anagrams of each other")
else:
    if sorted(string_1) != sorted(string_2):
        print(f"{string_1} and {string_2} are not anagrams of each other")
    else:
        print(f"{string_1} and {string_2} are anagrams of each other")
```
