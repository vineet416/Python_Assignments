# 4. Write a code to check if two given string are anagrams of each other.

```python
string_1 = input("Enter first word").lower()
string_2 = input("Enter second word").lower()

if len(string_1) != len(string_2):
    print(f"{string_1} and {string_2} are not anagrams of each other")
else:
    if sorted(string_1) != sorted(string_2):
        print(f"{string_1} and {string_2} are not anagrams of each other")
    else:
        print(f"{string_1} and {string_2} are anagrams of each other")
```

# 5. Write a code to find all occurrences of a given substring within another string.

```python
main_string = "hello, hello world, hello there"
substring = "hello"
start_index = 0
occurrences = []

while True:
    index = main_string.find(substring, start_index)
    if index == -1:
        break
    occurrences.append(index)
    start_index = index + 1

print(f"The substring '{substring}' occurs at indices: {occurrences}")
```

# 6. Write a code to perform basic string compression using of repeated characters.

```python
input_string = "aabcccccaaa"
compressed_string = ""
count = 1

for i in range(1, len(input_string)):
    if input_string[i] == input_string[i - 1]:
        count += 1
    else:
        compressed_string += input_string[i - 1] + str(count)
        count = 1

compressed_string += input_string[-1] + str(count)

if len(compressed_string) < len(input_string):
    print(compressed_string)
else:
    print(input_string)
```

# 14. Write a code to reverse a list in-place without using any built-in reverse functions.

```python
my_list = [1, 2, 3, 4, 5]
list_length = len(my_list)


for i in range(list_length // 2):
    my_list[i], my_list[list_length - 1 - i] = my_list[list_length - 1 - i], my_list[i]

print(my_list)
```

# 16. Create a code to check if a given list is sorted(either in ascending or descending order) or not.

```python
list1 = [1, 3, 5, 7, 9]
list2 = [2, 4, 6, 8, 10]

merged_list = []
i = 0
j = 0

while i < len(list1) and j < len(list2):
    if list1[i] < list2[j]:
        merged_list.append(list1[i])
        i += 1
    else:
        merged_list.append(list2[j])
        j += 1

merged_list.extend(list1[i:])
merged_list.extend(list2[j:])

print(merged_list)
```

# 20. Write a code to shuffle a given list randomly without using any built-in shuffle functions.

```python
import random

original_list = [1, 2, 3, 4, 5]
shuffled_list = original_list[:]

n = len(shuffled_list)
for i in range(n - 1, 0, -1):
    j = random.randint(0, i)
    shuffled_list[i], shuffled_list[j] = shuffled_list[j], shuffled_list[i]  # Swap elements

print(shuffled_list)
```

# 31. Write a code that takes a list of words as input and returns a dictionary where the keys are unique words and the values are the frequencies of those words in the input list.

```python
words_list = input("Enter a list of words separated by spaces: ").split()

word_freq_dict = {}

for word in words_list:
    if word in word_freq_dict:
        word_freq_dict[word] += 1
    else:
        word_freq_dict[word] = 1

print("Word frequencies:")
print(word_freq_dict)
```

# 32. Write a code that takes two dictionaries as input and merges them into a single dictionary. If there are common keys, the values should be added together.

```python
dict1 = {'a': 1, 'b': 2, 'c': 3}
dict2 = {'b': 3, 'c': 4, 'd': 5}

merged_dict = dict1.copy()

for key, value in dict2.items():
    if key in merged_dict:
        merged_dict[key] += value
    else:
        merged_dict[key] = value

print("Merged Dictionary:")
print(merged_dict)
```

# 33. Write a code to access a value in a nested dictionary. The function should take the dictionary and a list of keys as input, and return the corresponding value. If any keys do not exist in the dictionary, the function should return None.

```python
nested_dict = {
    'key1': {
        'key2': {
            'key3': 'value'
        }
    }
}

keys_list = ['key1', 'key2', 'key3']

value = nested_dict
for key in keys_list:
    if key in value:
        value = value[key]
    else:
        value = None
        break

print(value)
```

# 35. Write a code that inverts a dictionary, swapping keys and values. Ensure that the inverted dictionary correctly handles cases where multiple keys have the same value by storing the keys as a list in the inverted dictionary.

```python
original_dict = {'a': 1, 'b': 2, 'c': 1, 'd': 3, 'e': 2}

inverted_dict = {}

for key, value in original_dict.items():
    if value in inverted_dict:
        inverted_dict[value].append(key)  
    else:
        inverted_dict[value] = [key] 

print(inverted_dict)
```
